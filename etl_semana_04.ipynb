# Importar bibliotecas ou módulos

```python
import re
from datetime import datetime
```

```python
# construir função para converter o valor do arquivo vendas
# arrumar o valor com a exclusão do ponto e substituir a vírgula por ponto

def converter_valor(valor):
  # Se já estiver no formato internacional com ponto e sem milhar (ex: 1250.50)
  if "," not in valor and "." in valor:
    return float(valor)

  # Remove o ponto de milhar e troca a vírgula decimal por ponto
  valor = valor.replace(".", "").replace(",", ".")
  return float(valor)
                #vai retornar o valor alterado no arqueivo
```

```python
# Função para converter a data do relatório
def converter_data(data):
    return datetime.strptime(data, "%d/%m/%Y")
```

```python
# Função para processar cada linha do relatório

def processar_linha(linha):

    # para ignorar as linhas " Cancelado" do relatório
    if "CANCELADO" in linha:
        return None

    # Encontrar padrões 
    # (os valores de pedido, data, cliente, valor)
    padrao = (
        r"Pedido #(\d+)\s*\|\s*"
        r"Data=(\d{2}/\d{2}/\d{4})\s*\|\s*"
        r"Cliente:\s*(.*?)\s*\|\s*"
        r"Valor:\s*R\$\s*([\d.,]+)"
    )

    resultado = re.search(padrao, linha)

    if resultado is None:           # se ele encontrar ele vai continar, senãoele vai ignorar
        return None 

    pedido, data, cliente, valor = resultado.groups()

    return {
        "pedido": int(pedido),
        "data" : converter_data(data),
        "cliente": cliente,
        "valor": converter_valor(valor)
    }
```

# Fazer a leitura do arquivo 

```python
#colocar dentro de uma lista - para cada pedido ele vai colocr numa lista

vendas = []
with open("vendas_julho.txt", "r", encoding="utf-8") as arquivo:
    for linha in arquivo:
        venda = processar_linha(linha)
        if venda:
            vendas.append(venda)


```

```python
# Exibindo as vendas

for venda in vendas:
    print(
        venda["pedido"],
        venda["data"].strftime("%d/%m/%Y"),
        venda["cliente"],
        f"R$ {venda["valor"]: 2f}"
    )



```

```python
# Função para calcular o total de vendas do mês
def calcular_total(vendas):
    total = 0

    for venda in vendas:
        total += venda["valor"]

    return total
```

```python
# Exibir o valor total

print(f"Quantidade de vendas: {len(vendas)}")   #usar o len para definir a venda. Cada linha é uma venda. A quantidade de venda vai ser extraída pela qtidade de elementos da lista
print("Valor total vendido: R$ {calcular_total(vendas)}")
```
